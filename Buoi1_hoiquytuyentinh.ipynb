```python
from os import umask
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,r2_score
# tạo dữ liệu ngẫu nhiên
#với x là biến đầu vào và y là biến đầu ra
np.random.seed(0) #
x=2*np.random.rand(100,1)
y=4+3*x+np.random.rand(100,1)
#chia dữ liệu để thành tập huấn luyện và kiểm tra
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2 ,random_state=42 )
#tạo mô hình hòi quy tuyến tính
model =LinearRegression()
#huấn luyện mo hình trên tập huấn luyện
model.fit(x_train,y_train)
#dự đoán kết quả trên tập kiểm tra
y_pred=model.predict(x_test)
#đánh giá mô hình
mse=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
print("Mean Squared Error:",mse)
print("R-squared:",r2)
#in ra kết quả
print(f"hệ số chặn là (Intercept):{model.intercept_}")
print(f"hệ số góc là (Slope):{model.coef_}")
print(f"R^2 score:{r2}")
#vẽ mô hình biểu đồ
plt.scatter(x_test,y_test,color='blue',label='Dữ liệu kiểm tra')
plt.plot(x_test,y_pred,color='red',label="dự đoán hồi quy ")
plt.title("hoi quy tuyen tinh don gian ")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
